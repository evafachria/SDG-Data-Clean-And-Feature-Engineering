## 02. Data Cleaning

### Langkah 1: Pembersihan & Penyaringan Kolom Waktu (`tahun`)
* **Permasalahan**: Kolom `tahun` terdeteksi berjenis teks (`object`) dan mengandung data tahun di luar rentang studi kasus.
* **Tindakan**: Mengubah kolom `tahun` menjadi numerik bulat dan menyaring data agar hanya menyimpan tahun **2023, 2024, dan 2025** pada variabel baru `df_clean`.

````python
import pandas as pd
import numpy as np

# Menduplikasi data mentah asli agar tidak terganggu
df_clean = df.copy()

# 1. Mengubah kolom tahun menjadi numerik
df_clean['tahun'] = pd.to_numeric(df_clean['tahun'], errors='coerce')

# 2. Menyaring hanya tahun 2023, 2024, dan 2025
df_clean = df_clean[df_clean['tahun'].isin([2023, 2024, 2025])]

# Verifikasi hasil
print("Daftar tahun unik setelah disaring:", df_clean['tahun'].unique())
print(f"Dimensi data saat ini: {df_clean.shape[0]} baris")
````

### Langkah 2: Pembersihan Karakter Pengotor & Konversi Tipe Data Numerik
* **Permasalahan**: Kolom indikator penting bertipe `object` karena mengandung simbol `%`, `?`, dan teks `tidak tersedia`.
* **Tindakan**: Menghapus karakter-karakter pengotor tersebut, mengubah nilai non-numerik menjadi `NaN`, dan mengonversi tipe data kolom tersebut menjadi desimal (`float64`).

````python
# Daftar kolom numerik yang terdeteksi kotor
kolom_numerik_kotor = [
    'pertumbuhan_produktivitas_tenaga_kerja',
    'persentase_pekerja_informal',
    'tingkat_pengangguran',
    'persentase_pemuda_neet',
    'indeks_pembangunan_manusia',
    'tingkat_kemiskinan',
    'rata_rata_upah_bulanan',
    'jumlah_umkm'
]

# Membersihkan karakter pengotor satu per satu di setiap kolom
for col in kolom_numerik_kotor:
    # Konversi ke string, hapus %, ?, dan teks 'tidak tersedia', lalu hilangkan spasi di ujung
    df_clean[col] = df_clean[col].astype(str) \
        .str.replace('%', '', regex=False) \
        .str.replace('?', 'nan', regex=False) \
        .str.replace('tidak tersedia', 'nan', regex=False) \
        .str.strip()

    # Konversi ke numerik desimal (nilai yang tidak bisa diubah otomatis menjadi NaN)
    df_clean[col] = pd.to_numeric(df_clean[col], errors='coerce')

# Memeriksa hasil perubahan tipe data
print("Tipe data kolom setelah dibersihkan:")
print(df_clean[kolom_numerik_kotor].dtypes)
````

### Langkah 3: Pembersihan Koordinat Geografis (`latitude` & `longitude`)
* **Permasalahan**: Ditemukan nilai koordinat ekstrem tidak valid seperti `-999.0` dan `999.0` yang digunakan sebagai placeholder data kosong.
* **Tindakan**: Mengubah nilai koordinat di luar batas logis geografis (Latitude: -90 s.d 90, Longitude: -180 s.d 180) menjadi `NaN`.

````python
# Menampilkan statistik koordinat sebelum dibersihkan
print("Sebelum Pembersihan:")
print(df_clean[['latitude', 'longitude']].describe())

# Mengganti nilai di luar batas koordinat bumi yang valid dengan NaN
df_clean.loc[(df_clean['latitude'] < -90) | (df_clean['latitude'] > 90), 'latitude'] = np.nan
df_clean.loc[(df_clean['longitude'] < -180) | (df_clean['longitude'] > 180), 'longitude'] = np.nan

# Menampilkan statistik koordinat setelah dibersihkan
print(
"\nSetelah Pembersihan:")
print(df_clean[['latitude', 'longitude']].describe())
print(f"\nJumlah NaN pada latitude: {df_clean['latitude'].isna().sum()}")
print(f"Jumlah NaN pada longitude: {df_clean['longitude'].isna().sum()}")
````

### Langkah 4: Koreksi Nilai Tidak Logis pada Akses Internet
* **Permasalahan**: Kolom `persentase_akses_internet` terdeteksi memiliki nilai minimum negatif, yang secara logis tidak valid untuk data persentase.
* **Tindakan**: Mengubah semua nilai negatif pada kolom `persentase_akses_internet` menjadi `NaN`.

````python
# Menampilkan jumlah baris dengan nilai internet negatif sebelum koreksi
internet_negatif = df_clean[df_clean['persentase_akses_internet'] < 0]
print(f"Jumlah baris dengan nilai internet negatif: {len(internet_negatif)}")
if len(internet_negatif) > 0:
    print("Contoh data bernilai negatif:")
    print(internet_negatif[['nama_kabupaten_kota', 'tahun', 'persentase_akses_internet']])

# Mengubah nilai negatif menjadi NaN
df_clean.loc[df_clean['persentase_akses_internet'] < 0, 'persentase_akses_internet'] = np.nan

# Verifikasi nilai minimum setelah koreksi
print("\nStatistik kolom persentase_akses_internet setelah koreksi:")
print(df_clean['persentase_akses_internet'].describe())
````

### Langkah 4 (Tambahan): Koreksi Nilai Persentase Internet di Atas 100%
* **Permasalahan**: Nilai maksimum pada kolom `persentase_akses_internet` terdeteksi sebesar `130.0`, yang tidak logis untuk data persentase.
* **Tindakan**: Mengidentifikasi baris dengan persentase internet > 100% dan mengubahnya menjadi `NaN`.

````python
# Menampilkan baris yang memiliki persentase akses internet > 100
internet_over_100 = df_clean[df_clean['persentase_akses_internet'] > 100]
print(f"Jumlah baris dengan persentase internet > 100%: {len(internet_over_100)}")
if len(internet_over_100) > 0:
    print("\nContoh data internet > 100%:")
    print(internet_over_100[['nama_kabupaten_kota', 'tahun', 'persentase_akses_internet']])

# Mengubah nilai yang lebih dari 100 menjadi NaN
df_clean.loc[df_clean['persentase_akses_internet'] > 100, 'persentase_akses_internet'] = np.nan

# Verifikasi ulang statistik deskriptif kolom tersebut
print("\nStatistik kolom persentase_akses_internet setelah perbaikan akhir:")
print(df_clean['persentase_akses_internet'].describe())
````

### Langkah 5: Standardisasi Kategori Administrasi (`jenis_kabupaten_kota`)
* **Permasalahan**: Potensi adanya inkonsistensi penulisan kategori (seperti perbedaan huruf besar/kecil atau spasi tambahan) pada kolom klasifikasi wilayah.
* **Tindakan**: Memeriksa nilai unik pada kolom `jenis_kabupaten_kota`, menghapus spasi di ujung teks (*trimming*), dan menyeragamkan format penulisan menggunakan huruf kapital di awal kata (*Title Case*).

### Langkah 7: Imputasi Nilai Kosong Berjenjang (*Hierarchical Imputation*)

Untuk menjaga kualitas data, kita mengisi data kosong (`NaN`) menggunakan pendekatan hierarki:
1. **Rujukan Utama (Kabupaten/Kota yang Sama)**: Mengisi nilai kosong menggunakan nilai dari kabupaten/kota yang sama pada tahun yang berbeda (sangat penting untuk kestabilan koordinat geografis).
2. **Rujukan Cadangan 1 (Provinsi yang Sama)**: Jika kabupaten/kota tersebut kosong di semua tahun, kita gunakan rata-rata/median dari provinsi yang sama.
3. **Rujukan Cadangan 2 (Kelompok Pulau yang Sama)**: Sebagai jaring pengaman terakhir jika data provinsi pun tidak tersedia.

1. **Kelompok Koordinat Geografis (latitude & longitude)**
Karakteristik: Data numerik kontinu presisi tinggi yang melambangkan titik fisik wilayah di peta.
Mengapa tidak bisa dirata-rata/di-modus secara nasional? Jika kita mengisi koordinat Kabupaten Bogor yang kosong menggunakan rata-rata Indonesia, titik petanya bisa melompat ke tengah laut.
Solusi: Kita menggunakan Metode Lookup Historis. Posisi suatu daerah tidak pernah berpindah. Kita mencari data koordinat kabupaten/kota tersebut pada tahun lain di dalam dataset kita sendiri. Jika data tahun 2024 kosong, kita ambil koordinat dari data tahun 2023 atau 2025 miliknya.
2. **Kelompok Indikator Numerik Kontinu**
Variabel: rata_rata_upah_bulanan, persentase_akses_internet, jumlah_umkm, tingkat_pengangguran, persentase_pemuda_neet, persentase_pekerja_informal, pertumbuhan_produktivitas_tenaga_kerja, tingkat_kemiskinan, pertumbuhan_ekonomi, dan indeks_pembangunan_manusia (IPM).
Karakteristik: Data berupa angka desimal atau nominal yang memiliki sebaran dinamis.
Logika Pengisian (Mengapa Menggunakan Median, Bukan Mean?):
Median (nilai tengah) jauh lebih aman dari pencilan (outliers) dibanding rata-rata (mean). Contoh: Jika dalam suatu provinsi ada satu kota metropolitan dengan upah bulanan sangat tinggi (misalnya Rp 5.000.000) dan kabupaten lain berupah rendah (Rp 2.000.000), nilai rata-rata akan terdorong naik secara tidak adil. Median menjaga nilai tetap objektif mewakili daerah sekitar.
Jenjang Rujukan: Jadi di prioritaskan lookup ke tahun lain pada kabupaten yang sama terlebih dahulu. Jika kabupaten tersebut datanya kosong sama sekali di semua tahun, kita turun (fallback) menggunakan median provinsi tempat kabupaten tersebut berada.
3. **Kelompok Karakteristik Kategorikal**
Variabel: sektor_pekerjaan_dominan dan kategori_urbanisasi.
Karakteristik: Data berupa teks/kategori, bukan angka.
Logika Pengisian (Mengapa Menggunakan Modus?):
Data kategori tidak bisa dijumlahkan atau dibagi untuk dicari nilai tengahnya. Kita tidak bisa menghitung "rata-rata" dari sektor Pertanian dan Industri.
Modus (nilai yang paling sering muncul) adalah pilihan satu-satunya yang logis secara statistik untuk data kategorikal.
Jenjang Rujukan: Kita cari kategori dominan (modus) dari kabupaten yang sama di tahun lain. Jika benar-benar kosong di semua tahun, kita cari modus (sektor/kategori urbanisasi yang paling umum muncul) di provinsi atau kelompok pulau yang sama.

````python
import numpy as np
import pandas as pd

# Memastikan kita mengambil data bersih sebelum imputasi
if 'df_sebelum_imputasi' in locals() or 'df_sebelum_imputasi' in globals():
    df_clean = df_sebelum_imputasi.copy()
else:
    df_sebelum_imputasi = df_clean.copy()

# Daftar pemisahan jenis kolom
kolom_koordinat = ['latitude', 'longitude']
kolom_numerik_indikator = [
    'rata_rata_upah_bulanan', 'persentase_akses_internet', 'jumlah_umkm',
    'tingkat_pengangguran', 'persentase_pemuda_neet', 'persentase_pekerja_informal',
    'pertumbuhan_produktivitas_tenaga_kerja', 'tingkat_kemiskinan', 'pertumbuhan_ekonomi',
    'indeks_pembangunan_manusia'
]
kolom_kategorikal = ['sektor_pekerjaan_dominan', 'kategori_urbanisasi']

print("=== MENJALANKAN IMPUTASI KOREKSI PRESISI (ULANG) ===\n")

# 1. IMPUTASI KOORDINAT (Hanya menggunakan forward fill & backward fill dalam kode_kabupaten_kota yang sama)
print("-> Memproses koordinat spasial...")
for col in kolom_koordinat:
    df_clean[col] = df_clean.groupby('kode_kabupaten_kota')[col].transform(lambda x: x.ffill().bfill())

# 2. IMPUTASI NUMERIK INDIKATOR (Hierarki: Lookup -> Median Provinsi -> Median Pulau)
print("-> Memproses indikator numerik...")
for col in kolom_numerik_indikator:
    # Tahap A: Lookup kabupaten/kota di tahun lain
    df_clean[col] = df_clean.groupby('kode_kabupaten_kota')[col].transform(lambda x: x.fillna(x.median() if not x.dropna().empty else np.nan))
    # Tahap B: Fallback Provinsi
    if df_clean[col].isna().sum() > 0:
        df_clean[col] = df_clean.groupby('kode_provinsi')[col].transform(lambda x: x.fillna(x.median() if not x.dropna().empty else np.nan))
    # Tahap C: Fallback Pulau
    if df_clean[col].isna().sum() > 0:
        df_clean[col] = df_clean[col].fillna(df_clean.groupby('kelompok_pulau')[col].transform('median'))

# 3. IMPUTASI KATEGORIKAL (Hierarki: Lookup -> Modus Provinsi -> Modus Pulau)
print("-> Memproses atribut kategorikal...")
for col in kolom_kategorikal:
    # Tahap A: Lookup kabupaten/kota di tahun lain
    df_clean[col] = df_clean.groupby('kode_kabupaten_kota')[col].transform(lambda x: x.fillna(x.mode()[0] if not x.dropna().empty else np.nan))
    # Tahap B: Fallback Provinsi
    if df_clean[col].isna().sum() > 0:
        df_clean[col] = df_clean.groupby('kode_provinsi')[col].transform(lambda x: x.fillna(x.mode()[0] if not x.dropna().empty else np.nan))
    # Tahap C: Fallback Pulau
    if df_clean[col].isna().sum() > 0:
        df_clean[col] = df_clean[col].fillna(df_clean.groupby('kelompok_pulau')[col].transform(lambda x: x.mode()[0] if not x.dropna().empty else 'Tidak Diketahui'))

# Verifikasi Akhir
kolom_semua = kolom_koordinat + kolom_numerik_indikator + kolom_kategorikal
sisa_nan_df = df_clean[kolom_semua].isna().sum().to_frame(name='Jumlah Sisa NaN')

print("\n=== PROSES SELESAI ===")
display(sisa_nan_df)
````

### 3.5. Normalisasi Format `kode_kabupaten_kota`
Kita akan memperbaiki nilai kode kabupaten/kota yang tidak konsisten (misal: `1173.00` diubah menjadi `11.73`) agar struktur data terintegrasi dengan sempurna.

````python
import numpy as np

def normalisasi_kode(kode):
    if pd.isna(kode):
        return np.nan
    # Konversi ke string tanpa desimal tidak perlu
    kode_str = str(kode).strip()
    if '.' in kode_str:
        # Pisahkan angka sebelum dan setelah titik
        parts = kode_str.split('.')
        # Jika bagian kiri panjangnya 4 angka tanpa titik di tengah, contoh: 1173.0
        kiri = parts[0]
        if len(kiri) == 4:
            return f"{kiri[:2]}.{kiri[2:]}"
        elif len(kiri) == 2:
            return f"{kiri}.{parts[1][:2].ljust(2, '0')}"
    else:
        if len(kode_str) == 4:
            return f"{kode_str[:2]}.{kode_str[2:]}"
    return kode_str

# Terapkan fungsi normalisasi
df_clean['kode_kabupaten_kota'] = df_clean['kode_kabupaten_kota'].apply(normalisasi_kode)

# Verifikasi hasil penyatuan kode setelah normalisasi
duplikat_kunci_baru = df_clean.duplicated(subset=['kode_kabupaten_kota', 'tahun'], keep=False).sum()
print(f"Jumlah konflik kunci setelah normalisasi kode: {duplikat_kunci_baru}")

# Hitung ulang jumlah kombinasi unik
unik_kombinasi_baru = df_clean.groupby(['kode_kabupaten_kota', 'tahun']).size().reset_index(name='count')
print(f"Kombinasi unik saat ini: {len(unik_kombinasi_baru)} (Target: 1.542)")
````

### C. Indeks Komposit & Analisis Profil Wilayah (2 Kolom)

11. **`skor_risiko_sdm`**
    * **Tujuan**: Mengukur tingkat kerentanan modal manusia dalam kaitannya dengan produktivitas dan penyerapan kerja.
    * **Indikator Penyusun**: Tingkat Pengangguran Terbuka (TPT), Persentase Pemuda NEET, Persentase Pekerja Informal, dan Tingkat Kemiskinan.
    * **Metodologi Normalisasi**: Menggunakan standarisasi Min-Max agar rentang nilai berada pada skala 0 s.d 1. Karena seluruh indikator bersifat *cost* (semakin tinggi nilainya, semakin buruk risikonya), arah standarisasinya adalah searah:
      $$\text{Normalized Indicator} = \frac{X - X_{min}}{X_{max} - X_{min}}$$
    * **Pembobotan**: Pembobotan sama rata (*Equal Weighting*) masing-masing sebesar **25%**:
      $$\text{Skor Risiko SDM} = 0.25 \times (\text{Norm\_TPT} + \text{Norm\_NEET} + \text{Norm\_Informal} + \text{Norm\_Miskin})$$

12. **`profil_pembangunan_daerah`**
    * **Tujuan**: Segmentasi komprehensif wilayah untuk memetakan keselarasan antara pembangunan manusia, dinamika ekonomi, dan kesejahteraan dasar.
    * **Aturan Segmentasi (Eksplisit)**:
      * **`Maju & Sejahtera`**: Memiliki IPM Tinggi/Sangat Tinggi $(\ge 70.0)$, Pertumbuhan Ekonomi Positif $(> 0)$, dan tingkat kemiskinan di bawah median nasional tahun berjalan.
      * **`Pertumbuhan Rapuh`**: Memiliki Pertumbuhan Ekonomi Positif $(> 0)$, namun memiliki tingkat kemiskinan di atas median nasional tahun berjalan ATAU IPM masih tergolong Rendah/Sangat Rendah $(< 70.0)$.
      * **`Tertinggal / Rentan`**: Memiliki IPM Rendah/Sangat Rendah $(< 70.0)$ dan memiliki tingkat kemiskinan di atas median nasional tahun berjalan.
      * **`Krisis Pembangunan`**: Memiliki Pertumbuhan Ekonomi Negatif/Stagnan $(\le 0)$ disertai dengan tingkat kemiskinan di atas median nasional tahun berjalan.

````python
import pandas as pd
import numpy as np

# Memastikan df_clean terurut dengan baik berdasarkan wilayah dan tahun
df_clean = df_clean.sort_values(by=['kode_kabupaten_kota', 'tahun']).reset_index(drop=True)

# 1. Kategori IPM (Standar BPS)
def classify_ipm(ipm):
    if ipm < 60.0:
        return 'Sangat Rendah'
    elif ipm < 70.0:
        return 'Sedang'
    elif ipm < 80.0:
        return 'Tinggi'
    else:
        return 'Sangat Tinggi'

df_clean['kategori_ipm'] = df_clean['indeks_pembangunan_manusia'].apply(classify_ipm)

# 2. Kategori Pertumbuhan Ekonomi
def classify_growth(growth):
    if growth < 0:
        return 'Kontraksi'
    elif growth == 0:
        return 'Stagnan'
    else:
        return 'Tumbuh'

df_clean['kategori_pertumbuhan_ekonomi'] = df_clean['pertumbuhan_ekonomi'].apply(classify_growth)

# Fungsi bantu untuk pembagian tercil per tahun
def qcut_safe(series, labels=['Rendah', 'Sedang', 'Tinggi']):
    try:
        # Gunakan qcut dengan duplicates='drop' untuk keamanan jika ada nilai yang identik
        return pd.qcut(series, q=3, labels=labels, duplicates='drop')
    except Exception as e:
        # Fallback jika qcut gagal karena variansi data terlalu minim
        return pd.cut(series, bins=3, labels=labels)

# 3 s.d 7. Kategorisasi Tercil Tahunan (Annual Tertile Split)
for col_name, new_col in [
    ('tingkat_pengangguran', 'kategori_pengangguran'),
    ('tingkat_kemiskinan', 'kategori_kemiskinan'),
    ('persentase_pekerja_informal', 'kategori_informalitas'),
    ('persentase_akses_internet', 'kategori_akses_internet'),
    ('persentase_pemuda_neet', 'kategori_neet')
]:
    # Terapkan qcut berkelompok berdasarkan tahun
    df_clean[new_col] = df_clean.groupby('tahun')[col_name].transform(qcut_safe)
    # Pastikan tipe data kolom kategori dikonversi ke tipe string biasa
    df_clean[new_col] = df_clean[new_col].astype(str)

# Verifikasi dimensi dan sebaran hasil pengelompokan awal
print("=== VERIFIKASI SEBARAN FITUR GRUP A ===")
for cat_col in ['kategori_ipm', 'kategori_pertumbuhan_ekonomi', 'kategori_pengangguran']:
    print(f"\nSebaran untuk {cat_col}:")
    print(df_clean[cat_col].value_counts())

display(df_clean[['nama_kabupaten_kota', 'tahun', 'kategori_ipm', 'kategori_pertumbuhan_ekonomi', 'kategori_pengangguran', 'kategori_kemiskinan']].head(5))
````

````python
# Memastikan data terurut secara kronologis untuk setiap kabupaten/kota
df_clean = df_clean.sort_values(by=['kode_kabupaten_kota', 'tahun']).reset_index(drop=True)

# 8. perubahan_ipm_yoy
df_clean['perubahan_ipm_yoy'] = df_clean.groupby('kode_kabupaten_kota')['indeks_pembangunan_manusia'].diff()

# 9. perubahan_tpt_yoy
df_clean['perubahan_tpt_yoy'] = df_clean.groupby('kode_kabupaten_kota')['tingkat_pengangguran'].diff()

# 10. perubahan_kemiskinan_yoy
df_clean['perubahan_kemiskinan_yoy'] = df_clean.groupby('kode_kabupaten_kota')['tingkat_kemiskinan'].diff()

# Memastikan tahun 2023 bernilai NaN secara eksplisit
df_clean.loc[df_clean['tahun'] == 2023, ['perubahan_ipm_yoy', 'perubahan_tpt_yoy', 'perubahan_kemiskinan_yoy']] = np.nan

# Verifikasi hasil perhitungan YoY
print("=== VERIFIKASI PERHITUNGAN YoY (CONTOH DATA KAB. ACEH SELATAN) ===")
display(df_clean[df_clean['kode_kabupaten_kota'] == '11.01'][['nama_kabupaten_kota', 'tahun', 'indeks_pembangunan_manusia', 'perubahan_ipm_yoy', 'tingkat_pengangguran', 'perubahan_tpt_yoy', 'tingkat_kemiskinan', 'perubahan_kemiskinan_yoy']])

# Memeriksa jumlah nilai non-null per tahun untuk memastikan 2023 kosong sempurna
print("\n=== JUMLAH DATA NON-NULL YoY PER TAHUN ===")
print(df_clean.groupby('tahun')[['perubahan_ipm_yoy', 'perubahan_tpt_yoy', 'perubahan_kemiskinan_yoy']].count())
````

````python
# 11. Perhitungan skor_risiko_sdm (Komposit Min-Max dengan Bobot Sama 25%)
# Indikator risiko yang digunakan (semua bersifat cost / semakin tinggi semakin buruk)
risiko_cols = [
    'tingkat_pengangguran',
    'persentase_pemuda_neet',
    'persentase_pekerja_informal',
    'tingkat_kemiskinan'
]

# Lakukan normalisasi Min-Max berkelompok per tahun untuk menjaga keadilan perbandingan antartahun
for col in risiko_cols:
    min_val = df_clean.groupby('tahun')[col].transform('min')
    max_val = df_clean.groupby('tahun')[col].transform('max')
    df_clean[f'norm_{col}'] = (df_clean[col] - min_val) / (max_val - min_val)

# Hitung indeks komposit dengan rata-rata bobot setara (25% masing-masing)
df_clean['skor_risiko_sdm'] = df_clean[[
    'norm_tingkat_pengangguran',
    'norm_persentase_pemuda_neet',
    'norm_persentase_pekerja_informal',
    'norm_tingkat_kemiskinan'
]].mean(axis=1)

# Hapus kembali kolom normalisasi penolong agar dataframe tetap bersih
df_clean = df_clean.drop(columns=[f'norm_{col}' for col in risiko_cols])


# 12. Klasifikasi profil_pembangunan_daerah
# Hitung median kemiskinan nasional per tahun berjalan
df_clean['median_kemiskinan_tahunan'] = df_clean.groupby('tahun')['tingkat_kemiskinan'].transform('median')

def classify_profile(row):
    ipm = row['indeks_pembangunan_manusia']
    growth = row['pertumbuhan_ekonomi']
    poverty = row['tingkat_kemiskinan']
    med_poverty = row['median_kemiskinan_tahunan']

    if ipm >= 70.0 and growth > 0 and poverty < med_poverty:
        return 'Maju & Sejahtera'
    elif growth > 0 and (poverty > med_poverty or ipm < 70.0):
        return 'Pertumbuhan Rapuh'
    elif ipm < 70.0 and poverty > med_poverty:
        return 'Tertinggal / Rentan'
    elif growth <= 0 and poverty > med_poverty:
        return 'Krisis Pembangunan'
    else:
        # Fallback jika ada variasi di luar kondisi utama
        return 'Pertumbuhan Rapuh'

df_clean['profil_pembangunan_daerah'] = df_clean.apply(classify_profile, axis=1)

# Hapus kolom pembantu median kemiskinan
df_clean = df_clean.drop(columns=['median_kemiskinan_tahunan'])

# Verifikasi akhir seluruh 12 kolom baru
kolom_baru_12 = [
    'kategori_ipm', 'kategori_pertumbuhan_ekonomi', 'kategori_pengangguran',
    'kategori_kemiskinan', 'kategori_informalitas', 'kategori_akses_internet', 'kategori_neet',
    'perubahan_ipm_yoy', 'perubahan_tpt_yoy', 'perubahan_kemiskinan_yoy',
    'skor_risiko_sdm', 'profil_pembangunan_daerah'
]

print("=== VERIFIKASI SEBARAN PROFIL PEMBANGUNAN DAERAH ===")
print(df_clean['profil_pembangunan_daerah'].value_counts())

print("\n=== STATISTIK DESKRIPTIF SKOR RISIKO SDM ===")
print(df_clean['skor_risiko_sdm'].describe())

print("\n=== CONTOH HASIL AKHIR (12 KOLOM BARU TERBENTUK) ===")
display(df_clean[['nama_kabupaten_kota', 'tahun'] + kolom_baru_12].head(5))
````
