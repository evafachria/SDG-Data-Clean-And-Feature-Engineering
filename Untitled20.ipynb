## 02. Data Cleaning

### Langkah 1: Pembersihan & Penyaringan Kolom Waktu (`tahun`)
* **Permasalahan**: Kolom `tahun` terdeteksi berjenis teks (`object`) dan mengandung data tahun di luar rentang studi kasus.
* **Tindakan**: Mengubah kolom `tahun` menjadi numerik bulat dan menyaring data agar hanya menyimpan tahun **2023, 2024, dan 2025** pada variabel baru `df_clean`.

````python
import pandas as pd
import numpy as np

# Menduplikasi data mentah asli agar tidak terganggu
df_clean = df.copy()

# 1. Mengubah kolom tahun menjadi numerik
df_clean['tahun'] = pd.to_numeric(df_clean['tahun'], errors='coerce')

# 2. Menyaring hanya tahun 2023, 2024, dan 2025
df_clean = df_clean[df_clean['tahun'].isin([2023, 2024, 2025])]

# Verifikasi hasil
print("Daftar tahun unik setelah disaring:", df_clean['tahun'].unique())
print(f"Dimensi data saat ini: {df_clean.shape[0]} baris")
````

### Langkah 2: Pembersihan Karakter Pengotor & Konversi Tipe Data Numerik
* **Permasalahan**: Kolom indikator penting bertipe `object` karena mengandung simbol `%`, `?`, dan teks `tidak tersedia`.
* **Tindakan**: Menghapus karakter-karakter pengotor tersebut, mengubah nilai non-numerik menjadi `NaN`, dan mengonversi tipe data kolom tersebut menjadi desimal (`float64`).

````python

# ============================================================
# 3.3 KONVERSI TIPE DATA NUMERIK
# ============================================================

kolom_numerik_kotor = [
    'pertumbuhan_produktivitas_tenaga_kerja',
    'persentase_pekerja_informal',
    'tingkat_pengangguran',
    'persentase_pemuda_neet',
    'indeks_pembangunan_manusia',
    'tingkat_kemiskinan',
    'rata_rata_upah_bulanan',
    'jumlah_umkm'
]

# Membersihkan karakter pengotor
for col in kolom_numerik_kotor:

    nilai = (
        df_clean[col]
        .astype('string')
        .str.strip()
        .str.lower()
        .str.replace('%', '', regex=False)
        .str.strip()
    )

    # Mengubah placeholder menjadi missing value
    nilai = nilai.replace({
        '?': pd.NA,
        'tidak tersedia': pd.NA,
        'unknown': pd.NA,
        'null': pd.NA,
        'none': pd.NA,
        'nan': pd.NA,
        '-': pd.NA,
        '': pd.NA
    })

    # Konversi menjadi numerik
    df_clean[col] = pd.to_numeric(
        nilai,
        errors='coerce'
    )

# Verifikasi
print("=== HASIL KONVERSI TIPE DATA ===")
display(
    df_clean[kolom_numerik_kotor]
    .dtypes
    .to_frame('Tipe Data')
)

print("\n=== JUMLAH MISSING SETELAH KONVERSI ===")
display(
    df_clean[kolom_numerik_kotor]
    .isna()
    .sum()
    .to_frame('Jumlah Missing')
)
````

### Langkah 3: Pembersihan Koordinat Geografis (`latitude` & `longitude`)
* **Permasalahan**: Ditemukan nilai koordinat ekstrem tidak valid seperti `-999.0` dan `999.0` yang digunakan sebagai placeholder data kosong.
* **Tindakan**: Mengubah nilai koordinat di luar batas logis geografis (Latitude: -90 s.d 90, Longitude: -180 s.d 180) menjadi `NaN`.

````python

# ============================================================
# VALIDASI KOORDINAT GEOGRAFIS INDONESIA
# ============================================================

import numpy as np
import pandas as pd

# Menampilkan statistik sebelum pembersihan
print("=== SEBELUM PEMBERSIHAN ===")
display(df_clean[['latitude', 'longitude']].describe())

# Memastikan koordinat bertipe numerik
for col in ['latitude', 'longitude']:
    df_clean[col] = pd.to_numeric(
        df_clean[col], errors='coerce'
    )

# Batas geografis perkiraan Indonesia
batas_koordinat = {
    'latitude': (-11.5, 6.5),
    'longitude': (94.0, 142.0)
}

# Mengubah koordinat di luar batas menjadi NaN
for col, (batas_min, batas_max) in batas_koordinat.items():

    mask_invalid = (
        df_clean[col].notna()
        & ~df_clean[col].between(batas_min, batas_max)
    )

    print(f"{col}: {mask_invalid.sum()} nilai tidak valid")

    df_clean.loc[mask_invalid, col] = np.nan

# Menampilkan statistik setelah pembersihan
print("\n=== SETELAH PEMBERSIHAN ===")
display(df_clean[['latitude', 'longitude']].describe())

print("\n=== JUMLAH MISSING VALUES ===")
print(f"Latitude : {df_clean['latitude'].isna().sum()}")
print(f"Longitude: {df_clean['longitude'].isna().sum()}")
````

### Langkah 4: Koreksi Nilai Tidak Logis pada Akses Internet
* **Permasalahan**: Kolom `persentase_akses_internet` terdeteksi memiliki nilai minimum negatif, yang secara logis tidak valid untuk data persentase.
* **Tindakan**: Mengubah semua nilai negatif pada kolom `persentase_akses_internet` menjadi `NaN`.

````python

# ============================================================
# STANDARDISASI KODE DAN NAMA KABUPATEN/KOTA
# ============================================================

import pandas as pd

print("=== SEBELUM STANDARDISASI ===")
print("Wilayah unik:",
      df_clean['kode_kabupaten_kota'].nunique())

# Simpan kode asli untuk audit
df_clean['kode_kabupaten_kota_raw'] = (
    df_clean['kode_kabupaten_kota'].copy()
)

# Standardisasi kode wilayah
def normalisasi_kode(nilai):
    if pd.isna(nilai):
        return pd.NA

    teks = str(nilai).strip()

    # Format benar: 11.01, 32.13, dan sebagainya
    if pd.notna(teks) and __import__('re').fullmatch(
        r'\d{2}\.\d{2}', teks
    ):
        return teks

    # Format salah: 1101.00 atau 1101
    if __import__('re').fullmatch(r'\d{4}(?:\.0+)?', teks):
        angka = teks.split('.')[0]
        return angka[:2] + '.' + angka[2:]

    # Jangan menebak format yang tidak dikenali
    return pd.NA

df_clean['kode_kabupaten_kota'] = (
    df_clean['kode_kabupaten_kota']
    .apply(normalisasi_kode)
)

# Standardisasi nama wilayah
nama = (
    df_clean['nama_kabupaten_kota']
    .astype('string')
    .str.strip()
    .str.replace(r'\s+', ' ', regex=True)
    .str.replace(
        r'^Kab\.\s*',
        'Kabupaten ',
        regex=True,
        case=False
    )
    .str.replace(
        r'^Kot\.\s*',
        'Kota ',
        regex=True,
        case=False
    )
    .str.title()
)

df_clean['nama_kabupaten_kota'] = nama

print("\n=== SETELAH STANDARDISASI ===")
print("Wilayah unik:",
      df_clean['kode_kabupaten_kota'].nunique())

print("Kode wilayah NaN:",
      df_clean['kode_kabupaten_kota'].isna().sum())

print("Jumlah baris:",
      len(df_clean))

# Audit ulang
kunci = ['kode_kabupaten_kota', 'tahun']

print("\nBaris duplikat wilayah-tahun:",
      df_clean.duplicated(kunci, keep=False).sum())

print("\nKombinasi wilayah-tahun unik:",
      len(df_clean[kunci].drop_duplicates()))
````

````python

# ============================================================
# STANDARDISASI KODE WILAYAH - REVISI
# ============================================================

import pandas as pd
import re

# Gunakan kode asli yang disimpan sebelum standardisasi gagal
if 'kode_kabupaten_kota_raw' not in df_clean.columns:
    raise ValueError(
        "Kolom kode_kabupaten_kota_raw tidak ditemukan. "
        "Jalankan ulang notebook dari dataset RAW."
    )

def normalisasi_kode(nilai):
    if pd.isna(nilai):
        return pd.NA

    teks = str(nilai).strip()

    # Format empat digit: 1101 atau 1101.00
    if re.fullmatch(r'\d{4}(?:\.0+)?', teks):
        angka = teks.split('.')[0]
        return f"{angka[:2]}.{angka[2:]}"

    # Format standar: 11.01
    if re.fullmatch(r'\d{2}\.\d{2}', teks):
        return teks

    # Format numerik Excel: 11.1 -> 11.10
    # Catatan: hanya aman jika angka desimal memang
    # merupakan kode wilayah dua digit yang kehilangan nol akhir.
    if re.fullmatch(r'\d{2}\.\d', teks):
        return teks + '0'

    # Format 11.010 -> 11.01
    if re.fullmatch(r'\d{2}\.\d{2}0+', teks):
        return teks[:5]

    return pd.NA

# Kembalikan nilai asli sebelum normalisasi
df_clean['kode_kabupaten_kota'] = (
    df_clean['kode_kabupaten_kota_raw']
    .apply(normalisasi_kode)
)

# Verifikasi
print("=== HASIL STANDARDISASI REVISI ===")

print(
    "Wilayah unik:",
    df_clean['kode_kabupaten_kota'].nunique()
)

print(
    "Kode wilayah NaN:",
    df_clean['kode_kabupaten_kota'].isna().sum()
)

print(
    "Jumlah baris:",
    len(df_clean)
)

print(
    "Baris duplikat:",
    df_clean.duplicated(
        ['kode_kabupaten_kota', 'tahun'],
        keep=False
    ).sum()
)

print("\n=== KODE YANG BELUM BERHASIL DINORMALISASI ===")

display(
    df_clean.loc[
        df_clean['kode_kabupaten_kota'].isna(),
        [
            'kode_kabupaten_kota_raw',
            'nama_kabupaten_kota',
            'tahun'
        ]
    ]
)
````

### Langkah 7: Imputasi Nilai Kosong Berjenjang (*Hierarchical Imputation*)

Untuk menjaga kualitas data, kita mengisi data kosong (`NaN`) menggunakan pendekatan hierarki:
1. **Rujukan Utama (Kabupaten/Kota yang Sama)**: Mengisi nilai kosong menggunakan nilai dari kabupaten/kota yang sama pada tahun yang berbeda (sangat penting untuk kestabilan koordinat geografis).
2. **Rujukan Cadangan 1 (Provinsi yang Sama)**: Jika kabupaten/kota tersebut kosong di semua tahun, kita gunakan rata-rata/median dari provinsi yang sama.
3. **Rujukan Cadangan 2 (Kelompok Pulau yang Sama)**: Sebagai jaring pengaman terakhir jika data provinsi pun tidak tersedia.

1. **Kelompok Koordinat Geografis (latitude & longitude)**
Karakteristik: Data numerik kontinu presisi tinggi yang melambangkan titik fisik wilayah di peta.
Mengapa tidak bisa dirata-rata/di-modus secara nasional? Jika kita mengisi koordinat Kabupaten Bogor yang kosong menggunakan rata-rata Indonesia, titik petanya bisa melompat ke tengah laut.
Solusi: Kita menggunakan Metode Lookup Historis. Posisi suatu daerah tidak pernah berpindah. Kita mencari data koordinat kabupaten/kota tersebut pada tahun lain di dalam dataset kita sendiri. Jika data tahun 2024 kosong, kita ambil koordinat dari data tahun 2023 atau 2025 miliknya.
2. **Kelompok Indikator Numerik Kontinu**
Variabel: rata_rata_upah_bulanan, persentase_akses_internet, jumlah_umkm, tingkat_pengangguran, persentase_pemuda_neet, persentase_pekerja_informal, pertumbuhan_produktivitas_tenaga_kerja, tingkat_kemiskinan, pertumbuhan_ekonomi, dan indeks_pembangunan_manusia (IPM).
Karakteristik: Data berupa angka desimal atau nominal yang memiliki sebaran dinamis.
Logika Pengisian (Mengapa Menggunakan Median, Bukan Mean?):
Median (nilai tengah) jauh lebih aman dari pencilan (outliers) dibanding rata-rata (mean). Contoh: Jika dalam suatu provinsi ada satu kota metropolitan dengan upah bulanan sangat tinggi (misalnya Rp 5.000.000) dan kabupaten lain berupah rendah (Rp 2.000.000), nilai rata-rata akan terdorong naik secara tidak adil. Median menjaga nilai tetap objektif mewakili daerah sekitar.
Jenjang Rujukan: Jadi di prioritaskan lookup ke tahun lain pada kabupaten yang sama terlebih dahulu. Jika kabupaten tersebut datanya kosong sama sekali di semua tahun, kita turun (fallback) menggunakan median provinsi tempat kabupaten tersebut berada.
3. **Kelompok Karakteristik Kategorikal**
Variabel: sektor_pekerjaan_dominan dan kategori_urbanisasi.
Karakteristik: Data berupa teks/kategori, bukan angka.
Logika Pengisian (Mengapa Menggunakan Modus?):
Data kategori tidak bisa dijumlahkan atau dibagi untuk dicari nilai tengahnya. Kita tidak bisa menghitung "rata-rata" dari sektor Pertanian dan Industri.
Modus (nilai yang paling sering muncul) adalah pilihan satu-satunya yang logis secara statistik untuk data kategorikal.
Jenjang Rujukan: Kita cari kategori dominan (modus) dari kabupaten yang sama di tahun lain. Jika benar-benar kosong di semua tahun, kita cari modus (sektor/kategori urbanisasi yang paling umum muncul) di provinsi atau kelompok pulau yang sama.

````python

# ============================================================
# LANGKAH 7 - HIERARCHICAL IMPUTATION (FINAL LENGKAP)
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. SALIN DATA
# ------------------------------------------------------------

df_awal = df_clean.copy(deep=True)
df_hasil = df_awal.copy(deep=True)

kunci = ['kode_kabupaten_kota', 'tahun']

# ------------------------------------------------------------
# 2. KELOMPOK VARIABEL
# ------------------------------------------------------------

kolom_koordinat = [
    'latitude',
    'longitude'
]

kolom_numerik = [
    'rata_rata_upah_bulanan',
    'persentase_akses_internet',
    'jumlah_umkm',
    'tingkat_pengangguran',
    'persentase_pemuda_neet',
    'persentase_pekerja_informal',
    'pertumbuhan_produktivitas_tenaga_kerja',
    'tingkat_kemiskinan',
    'pertumbuhan_ekonomi',
    'indeks_pembangunan_manusia',
    'efisiensi_penggunaan_sumber_daya',
    'persentase_pekerja_anak',
    'indeks_keselamatan_kerja',
    'kontribusi_lapangan_kerja_pariwisata',
    'tingkat_partisipasi_angkatan_kerja',
    'tingkat_inklusi_keuangan'
]

kolom_kategorikal = [
    'sektor_pekerjaan_dominan',
    'kategori_urbanisasi',
    'strategi_ketenagakerjaan_pemuda',
    'prioritas_kebijakan',
    'sektor_ekonomi_dominan',
    'bantuan_perdagangan'
]

kolom_identitas = [
    'kode_provinsi',
    'nama_provinsi',
    'nama_kabupaten_kota',
    'jenis_kabupaten_kota',
    'kelompok_pulau'
]

kolom_raw = [
    'kode_kabupaten_kota_raw'
]

kolom_diperiksa = list(dict.fromkeys(
    kolom_koordinat
    + kolom_numerik
    + kolom_kategorikal
    + kolom_identitas
    + kolom_raw
))

kolom_wajib = list(dict.fromkeys(
    kunci + kolom_diperiksa
))

tidak_ada = [
    c for c in kolom_wajib
    if c not in df_hasil.columns
]

if tidak_ada:
    raise ValueError(
        f"Kolom tidak ditemukan: {tidak_ada}"
    )

# Validasi struktur sebelum imputasi
assert len(df_hasil) == 1542, "Jumlah baris bukan 1542."
assert df_hasil['kode_kabupaten_kota'].nunique() == 514
assert not df_hasil.duplicated(kunci).any()

# ------------------------------------------------------------
# 3. FUNGSI MODUS
# ------------------------------------------------------------

def modus_aman(x):
    hasil = x.dropna().mode()
    return hasil.iloc[0] if not hasil.empty else np.nan

# Simpan penanda nilai yang awalnya kosong
mask_missing_awal = {
    col: df_hasil[col].isna().copy()
    for col in kolom_diperiksa
}

# Urutkan berdasarkan wilayah dan tahun
df_hasil = df_hasil.sort_values(kunci).copy()

# ------------------------------------------------------------
# 4. IDENTITAS WILAYAH
# ------------------------------------------------------------

print("=== IMPUTASI IDENTITAS WILAYAH ===")

for col in kolom_identitas:

    referensi = (
        df_hasil.groupby('kode_kabupaten_kota')[col]
        .transform(modus_aman)
    )

    df_hasil[col] = df_hasil[col].fillna(referensi)

# ------------------------------------------------------------
# 5. KOORDINAT - HISTORICAL LOOKUP
# ------------------------------------------------------------

print("\n=== IMPUTASI KOORDINAT ===")

for col in kolom_koordinat:

    df_hasil[col] = pd.to_numeric(
        df_hasil[col],
        errors='coerce'
    ).astype('float64')

    df_hasil[col] = (
        df_hasil.groupby('kode_kabupaten_kota')[col]
        .transform(lambda x: x.ffill().bfill())
    )

# ------------------------------------------------------------
# 6. IMPUTASI NUMERIK
# Median wilayah -> provinsi -> pulau
# ------------------------------------------------------------

print("\n=== IMPUTASI 16 INDIKATOR NUMERIK ===")

for col in kolom_numerik:

    df_hasil[col] = pd.to_numeric(
        df_hasil[col],
        errors='coerce'
    ).astype('float64')

    for grup in [
        'kode_kabupaten_kota',
        'kode_provinsi',
        'kelompok_pulau'
    ]:

        median_grup = (
            df_hasil.groupby(grup)[col]
            .transform('median')
        )

        df_hasil[col] = df_hasil[col].fillna(
            median_grup
        )

# Jumlah UMKM berupa bilangan bulat
df_hasil['jumlah_umkm'] = (
    df_hasil['jumlah_umkm'].round()
)

# ------------------------------------------------------------
# 7. IMPUTASI KATEGORIKAL
# Modus wilayah -> provinsi -> pulau
# Termasuk bantuan_perdagangan
# ------------------------------------------------------------

print("\n=== IMPUTASI 6 VARIABEL KATEGORIKAL ===")

for col in kolom_kategorikal:

    for grup in [
        'kode_kabupaten_kota',
        'kode_provinsi',
        'kelompok_pulau'
    ]:

        modus_grup = (
            df_hasil.groupby(grup)[col]
            .transform(modus_aman)
        )

        df_hasil[col] = df_hasil[col].fillna(
            modus_grup
        )

# ------------------------------------------------------------
# 8. IMPUTASI KODE RAW
# Historical lookup kabupaten/kota yang sama
# ------------------------------------------------------------

print("\n=== IMPUTASI KODE KABUPATEN/KOTA RAW ===")

col = 'kode_kabupaten_kota_raw'

df_hasil[col] = df_hasil[col].astype('string')

df_hasil[col] = (
    df_hasil.groupby('kode_kabupaten_kota')[col]
    .transform(lambda x: x.ffill().bfill())
)

# Penanda nilai RAW hasil imputasi
df_hasil['kode_raw_diimputasi'] = (
    mask_missing_awal[col]
    & df_hasil[col].notna()
)

# ------------------------------------------------------------
# 9. LAPORAN IMPUTASI
# ------------------------------------------------------------

laporan_imputasi = pd.DataFrame({
    'Missing Sebelum':
        df_awal[kolom_diperiksa].isna().sum(),
    'Missing Sesudah':
        df_hasil[kolom_diperiksa].isna().sum()
})

laporan_imputasi['Berhasil Diisi'] = (
    laporan_imputasi['Missing Sebelum']
    - laporan_imputasi['Missing Sesudah']
)

# ------------------------------------------------------------
# 10. VALIDASI STRUKTUR
# ------------------------------------------------------------

assert len(df_hasil) == 1542
assert df_hasil['kode_kabupaten_kota'].nunique() == 514
assert not df_hasil.duplicated(kunci).any()

assert (
    df_hasil.groupby('kode_kabupaten_kota')['tahun']
    .nunique().eq(3).all()
), "Masih ada wilayah dengan tahun tidak lengkap."

# ------------------------------------------------------------
# 11. SIMPAN HASIL
# ------------------------------------------------------------

df_sebelum_imputasi = df_awal
df_clean = df_hasil.reset_index(drop=True)

# ------------------------------------------------------------
# 12. LAPORAN AKHIR
# ------------------------------------------------------------

print("\n=== LAPORAN HASIL IMPUTASI ===")
display(laporan_imputasi)

print("\n=== VERIFIKASI STRUKTUR ===")
print("Jumlah baris:", len(df_clean))
print(
    "Jumlah wilayah:",
    df_clean['kode_kabupaten_kota'].nunique()
)
print(
    "Duplikat wilayah-tahun:",
    df_clean.duplicated(kunci).sum()
)

print("\n=== SISA MISSING SELURUH DATASET ===")

sisa_missing = df_clean.isna().sum()

display(
    sisa_missing[sisa_missing > 0]
    .sort_values(ascending=False)
    .to_frame('Jumlah Missing')
)

print("\n=== STATUS IMPUTASI KODE RAW ===")
print(
    "Jumlah kode RAW hasil imputasi:",
    df_clean['kode_raw_diimputasi'].sum()
)

print("\n=== KESIMPULAN ===")

if sisa_missing.sum() == 0:
    print("SELURUH MISSING VALUES BERHASIL DITANGANI.")
else:
    print("Masih ada missing values yang perlu diperiksa.")

print("VALIDASI STRUKTUR: LULUS")
print("Langkah 7 selesai.")
````

## 04. Feature Engineering (12 Kolom Tambahan)

Pada tahap *Feature Engineering* ini, kita akan membentuk **12 kolom (fitur) baru** yang dikelompokkan ke dalam beberapa dimensi analisis. Fitur-fitur ini dirancang untuk mempermudah visualisasi data di Tableau serta mempertajam analisis pengelompokan wilayah prioritas pembangunan berdasarkan target SDG 8.

Berikut adalah rincian aturan, dasar klasifikasi, serta rumus perhitungan untuk masing-masing kolom baru:

### A. Kategorisasi Berbasis Ambang Batas Standard & Distribusi (7 Kolom)
Pengelompokan ini membantu mengonversi data numerik kontinu menjadi kategori diskrit untuk analisis demografi wilayah.

1. **`kategori_ipm`**
   * **Dasar Klasifikasi**: Klasifikasi resmi Badan Pusat Statistik (BPS) Republik Indonesia.
   * **Aturan Logika**:
     * $\text{IPM} < 60.0$: `Sangat Rendah`
     * $60.0 \le \text{IPM} < 70.0$: `Sedang`
     * $70.0 \le \text{IPM} < 80.0$: `Tinggi`
     * $\text{IPM} \ge 80.0$: `Sangat Tinggi`

2. **`kategori_pertumbuhan_ekonomi`**
   * **Dasar Klasifikasi**: Kinerja pertumbuhan ekonomi riil wilayah.
   * **Aturan Logika**:
     * $\text{Pertumbuhan} < 0$: `Kontraksi`
     * $\text{Pertumbuhan} == 0$: `Stagnan`
     * $\text{Pertumbuhan} > 0$: `Tumbuh`

3. **`kategori_pengangguran`**
   * **Dasar Klasifikasi**: Distribusi nilai Tingkat Pengangguran Terbuka (TPT) secara tahunan (*Annual Tertile Split*).
   * **Aturan Logika**: Nilai dipisahkan menjadi 3 kelompok (Tercil) dengan frekuensi sama setiap tahunnya:
     * Tercil 1 (Rendah)
     * Tercil 2 (Sedang)
     * Tercil 3 (Tinggi)

4. **`kategori_kemiskinan`**
   * **Dasar Klasifikasi**: Distribusi persentase penduduk miskin secara tahunan (*Annual Tertile Split*).
   * **Aturan Logika**:
     * Tercil 1 (Rendah)
     * Tercil 2 (Sedang)
     * Tercil 3 (Tinggi)

5. **`kategori_informalitas`**
   * **Dasar Klasifikasi**: Distribusi proporsi pekerja informal secara tahunan (*Annual Tertile Split*).
   * **Aturan Logika**:
     * Tercil 1 (Rendah)
     * Tercil 2 (Sedang)
     * Tercil 3 (Tinggi)

6. **`kategori_akses_internet`**
   * **Dasar Klasifikasi**: Distribusi persentase akses internet rumah tangga secara tahunan (*Annual Tertile Split*).
   * **Aturan Logika**:
     * Tercil 1 (Rendah)
     * Tercil 2 (Sedang)
     * Tercil 3 (Tinggi)

7. **`kategori_neet`**
   * **Dasar Klasifikasi**: Distribusi persentase pemuda *Not in Education, Employment, or Training* (NEET) secara tahunan (*Annual Tertile Split*).
   * **Aturan Logika**:
     * Tercil 1 (Rendah)
     * Tercil 2 (Sedang)
     * Tercil 3 (Tinggi)

---

### B. Indikator Perubahan Tahunan (Year-on-Year / YoY) (3 Kolom)
Indikator ini mengukur laju dinamika perkembangan wilayah. Karena keterbatasan baseline data historis, **tahun 2023 diatur secara eksplisit bernilai kosong (`NaN`)**.

8. **`perubahan_ipm_yoy`**
   $$\text{Perubahan IPM}_{t} = \text{IPM}_{t} - \text{IPM}_{t-1}$$ (dalam satuan poin indeks)

9. **`perubahan_tpt_yoy`**
   $$\text{Perubahan TPT}_{t} = \text{TPT}_{t} - \text{TPT}_{t-1}$$ (dalam satuan poin persentase)

10. **`perubahan_kemiskinan_yoy`**
    $$\text{Perubahan Kemiskinan}_{t} = \text{Kemiskinan}_{t} - \text{Kemiskinan}_{t-1}$$ (dalam satuan poin persentase)

---

### C. Indeks Komposit & Analisis Profil Wilayah (2 Kolom)

11. **`skor_risiko_sdm`**
    * **Tujuan**: Mengukur tingkat kerentanan modal manusia dalam kaitannya dengan produktivitas dan penyerapan kerja.
    * **Indikator Penyusun**: Tingkat Pengangguran Terbuka (TPT), Persentase Pemuda NEET, Persentase Pekerja Informal, dan Tingkat Kemiskinan.
    * **Metodologi Normalisasi**: Menggunakan standarisasi Min-Max agar rentang nilai berada pada skala 0 s.d 1. Karena seluruh indikator bersifat *cost* (semakin tinggi nilainya, semakin buruk risikonya), arah standarisasinya adalah searah:
      $$\text{Normalized Indicator} = \frac{X - X_{min}}{X_{max} - X_{min}}$$
    * **Pembobotan**: Pembobotan sama rata (*Equal Weighting*) masing-masing sebesar **25%**:
      $$\text{Skor Risiko SDM} = 0.25 \times (\text{Norm\_TPT} + \text{Norm\_NEET} + \text{Norm\_Informal} + \text{Norm\_Miskin})$$

12. **`profil_pembangunan_daerah`**
    * **Tujuan**: Segmentasi komprehensif wilayah untuk memetakan keselarasan antara pembangunan manusia, dinamika ekonomi, dan kesejahteraan dasar.
    * **Aturan Segmentasi (Eksplisit)**:
      * **`Maju & Sejahtera`**: Memiliki IPM Tinggi/Sangat Tinggi $(\ge 70.0)$, Pertumbuhan Ekonomi Positif $(> 0)$, dan tingkat kemiskinan di bawah median nasional tahun berjalan.
      * **`Pertumbuhan Rapuh`**: Memiliki Pertumbuhan Ekonomi Positif $(> 0)$, namun memiliki tingkat kemiskinan di atas median nasional tahun berjalan ATAU IPM masih tergolong Rendah/Sangat Rendah $(< 70.0)$.
      * **`Tertinggal / Rentan`**: Memiliki IPM Rendah/Sangat Rendah $(< 70.0)$ dan memiliki tingkat kemiskinan di atas median nasional tahun berjalan.
      * **`Krisis Pembangunan`**: Memiliki Pertumbuhan Ekonomi Negatif/Stagnan $(\le 0)$ disertai dengan tingkat kemiskinan di atas median nasional tahun berjalan.

````python

# ============================================================
# 04. FEATURE ENGINEERING - 12 KOLOM TAMBAHAN
# ============================================================

import pandas as pd
import numpy as np

# Salinan agar hasil cleaning tetap aman
df_fe = df_clean.copy(deep=True)

# Urutkan data panel
df_fe = df_fe.sort_values(
    ['kode_kabupaten_kota', 'tahun']
).reset_index(drop=True)

# Pastikan kunci panel unik
assert not df_fe.duplicated(
    ['kode_kabupaten_kota', 'tahun']
).any(), "Masih ada duplikat wilayah-tahun."

# ============================================================
# A. KATEGORISASI INDIKATOR (7 KOLOM)
# ============================================================

# 1. KATEGORI IPM
def classify_ipm(ipm):
    if pd.isna(ipm):
        return np.nan
    elif ipm < 60:
        return 'Sangat Rendah'
    elif ipm < 70:
        return 'Sedang'
    elif ipm < 80:
        return 'Tinggi'
    else:
        return 'Sangat Tinggi'

df_fe['kategori_ipm'] = (
    df_fe['indeks_pembangunan_manusia']
    .apply(classify_ipm)
)

# 2. KATEGORI PERTUMBUHAN EKONOMI
def classify_growth(x):
    if pd.isna(x):
        return np.nan
    elif x < 0:
        return 'Kontraksi'
    elif x == 0:
        return 'Stagnan'
    else:
        return 'Tumbuh'

df_fe['kategori_pertumbuhan_ekonomi'] = (
    df_fe['pertumbuhan_ekonomi']
    .apply(classify_growth)
)

# 3-7. KATEGORI TERCIL PER TAHUN
# Nilai identik dipertahankan pada kategori yang sama

def tercil_tahunan(series):
    hasil = pd.Series(
        pd.NA,
        index=series.index,
        dtype='string'
    )

    nilai = pd.to_numeric(series, errors='coerce')
    valid = nilai.notna()

    if valid.sum() == 0:
        return hasil

    peringkat = nilai[valid].rank(
        method='average',
        pct=True
    )

    hasil.loc[valid] = np.select(
        [
            peringkat <= 1/3,
            peringkat <= 2/3
        ],
        ['Rendah', 'Sedang'],
        default='Tinggi'
    )

    return hasil

fitur_tercil = {
    'tingkat_pengangguran':
        'kategori_pengangguran',
    'tingkat_kemiskinan':
        'kategori_kemiskinan',
    'persentase_pekerja_informal':
        'kategori_informalitas',
    'persentase_akses_internet':
        'kategori_akses_internet',
    'persentase_pemuda_neet':
        'kategori_neet'
}

for asal, baru in fitur_tercil.items():
    df_fe[baru] = (
        df_fe.groupby('tahun')[asal]
        .transform(tercil_tahunan)
    )

# ============================================================
# B. PERUBAHAN TAHUNAN / YoY (3 KOLOM)
# ============================================================

fitur_yoy = {
    'indeks_pembangunan_manusia':
        'perubahan_ipm_yoy',
    'tingkat_pengangguran':
        'perubahan_tpt_yoy',
    'tingkat_kemiskinan':
        'perubahan_kemiskinan_yoy'
}

for asal, baru in fitur_yoy.items():

    df_fe[baru] = (
        df_fe.groupby('kode_kabupaten_kota')[asal]
        .diff()
    )

    # Tahun 2023 tidak memiliki baseline 2022
    df_fe.loc[
        df_fe['tahun'] == 2023, baru
    ] = np.nan

# ============================================================
# C. INDEKS KOMPOSIT & PROFIL (2 KOLOM)
# ============================================================

# 11. SKOR RISIKO SDM
# Normalisasi Min-Max seluruh periode 2023-2025
# Bobot sama rata 25%

indikator_risiko = [
    'tingkat_pengangguran',
    'persentase_pemuda_neet',
    'persentase_pekerja_informal',
    'tingkat_kemiskinan'
]

kolom_normalisasi = []

for col in indikator_risiko:

    nilai = pd.to_numeric(
        df_fe[col], errors='coerce'
    )

    nilai_min = nilai.min()
    nilai_max = nilai.max()

    nama_norm = '_norm_' + col

    if nilai_max > nilai_min:
        df_fe[nama_norm] = (
            (nilai - nilai_min)
            / (nilai_max - nilai_min)
        )
    else:
        # Indikator konstan tidak membedakan risiko
        df_fe[nama_norm] = pd.Series(
            np.where(nilai.notna(), 0.0, np.nan),
            index=df_fe.index
        )

    kolom_normalisasi.append(nama_norm)

df_fe['skor_risiko_sdm'] = (
    df_fe[kolom_normalisasi].sum(
        axis=1,
        min_count=len(kolom_normalisasi)
    ) * 0.25
)

# Hapus kolom bantu agar tepat 12 fitur baru
df_fe = df_fe.drop(columns=kolom_normalisasi)

# 12. PROFIL PEMBANGUNAN DAERAH

# Median kemiskinan nasional per tahun
median_kemiskinan_tahunan = (
    df_fe.groupby('tahun')['tingkat_kemiskinan']
    .transform('median')
)

ipm = df_fe['indeks_pembangunan_manusia']
growth = df_fe['pertumbuhan_ekonomi']
poverty = df_fe['tingkat_kemiskinan']

# Urutan prioritas mengatasi aturan yang tumpang tindih
kondisi = [
    # Prioritas 1: Krisis Pembangunan
    (growth <= 0) &
    (poverty > median_kemiskinan_tahunan),

    # Prioritas 2: Tertinggal / Rentan
    (ipm < 70) &
    (poverty > median_kemiskinan_tahunan),

    # Prioritas 3: Maju & Sejahtera
    (ipm >= 70) &
    (growth > 0) &
    (poverty < median_kemiskinan_tahunan),

    # Prioritas 4: Pertumbuhan Rapuh
    (growth > 0) &
    (
        (poverty > median_kemiskinan_tahunan)
        | (ipm < 70)
    )
]

pilihan = [
    'Krisis Pembangunan',
    'Tertinggal / Rentan',
    'Maju & Sejahtera',
    'Pertumbuhan Rapuh'
]

df_fe['profil_pembangunan_daerah'] = (
    np.select(
        kondisi,
        pilihan,
        default='Belum Terklasifikasi'
    )
)

# ============================================================
# D. VERIFIKASI 12 FITUR
# ============================================================

fitur_baru = [
    'kategori_ipm',
    'kategori_pertumbuhan_ekonomi',
    'kategori_pengangguran',
    'kategori_kemiskinan',
    'kategori_informalitas',
    'kategori_akses_internet',
    'kategori_neet',
    'perubahan_ipm_yoy',
    'perubahan_tpt_yoy',
    'perubahan_kemiskinan_yoy',
    'skor_risiko_sdm',
    'profil_pembangunan_daerah'
]

print("=== VERIFIKASI FEATURE ENGINEERING ===")

print("Jumlah fitur tambahan:", len(fitur_baru))
print("Jumlah baris:", len(df_fe))
print("Jumlah wilayah:",
      df_fe['kode_kabupaten_kota'].nunique())

print("\n=== MISSING VALUES FITUR BARU ===")
display(
    df_fe[fitur_baru]
    .isna().sum()
    .to_frame('Jumlah Missing')
)

print("\n=== DISTRIBUSI PROFIL PEMBANGUNAN ===")
display(
    pd.crosstab(
        df_fe['tahun'],
        df_fe['profil_pembangunan_daerah']
    )
)

print("\n=== STATISTIK SKOR RISIKO SDM ===")
display(
    df_fe['skor_risiko_sdm']
    .describe()
    .to_frame('Skor Risiko SDM')
)

print("\n=== CONTOH 12 FITUR BARU ===")
display(
    df_fe[
        ['nama_kabupaten_kota', 'tahun']
        + fitur_baru
    ].head(10)
)

# ============================================================
# E. VALIDASI DAN SIMPAN
# ============================================================

assert len(fitur_baru) == 12
assert len(df_fe) == len(df_clean)

assert not df_fe.duplicated(
    ['kode_kabupaten_kota', 'tahun']
).any()

# Tahun 2023 harus kosong pada fitur YoY
for col in fitur_yoy.values():
    assert df_fe.loc[
        df_fe['tahun'] == 2023, col
    ].isna().all()

# Simpan hasil
df_sebelum_feature_engineering = df_clean.copy(
    deep=True
)

df_clean = df_fe.copy(deep=True)

print("\nFEATURE ENGINEERING 12 KOLOM SELESAI.")
````

````python

import pandas as pd
import numpy as np

# ============================================================
# 3.11 AUDIT MENYELURUH HASIL DATA CLEANING
# ============================================================

data_audit = df_clean.copy()

hasil_audit = []
detail_masalah = []

kunci = ['kode_kabupaten_kota', 'tahun']
tahun_target = {2023, 2024, 2025}

def catat(nama, mask, tingkat='GAGAL', keterangan=''):
    """Mencatat jumlah dan lokasi baris bermasalah."""
    mask = pd.Series(mask, index=data_audit.index)
    mask = mask.fillna(False).astype(bool)

    jumlah = int(mask.sum())

    hasil_audit.append({
        'Pemeriksaan': nama,
        'Status': (
            'LULUS' if jumlah == 0 else tingkat
        ),
        'Jumlah Masalah': jumlah,
        'Keterangan': keterangan
    })

    if jumlah:
        baris = data_audit.loc[mask, kunci].copy()
        baris['Baris DataFrame'] = baris.index
        baris['Masalah'] = nama
        baris['Tingkat'] = tingkat
        detail_masalah.append(baris)


# ============================================================
# A. AUDIT STRUKTUR DATA
# ============================================================

print("A. AUDIT STRUKTUR DATA")

hasil_audit.extend([
    {
        'Pemeriksaan': 'Jumlah baris = 1542',
        'Status': 'LULUS' if len(data_audit) == 1542 else 'GAGAL',
        'Jumlah Masalah': abs(len(data_audit) - 1542),
        'Keterangan': f'Aktual: {len(data_audit)}'
    },
    {
        'Pemeriksaan': 'Jumlah wilayah = 514',
        'Status': 'LULUS' if data_audit[
            'kode_kabupaten_kota'
        ].nunique() == 514 else 'GAGAL',
        'Jumlah Masalah': abs(
            data_audit['kode_kabupaten_kota'].nunique() - 514
        ),
        'Keterangan': '514 wilayah unik'
    }
])

catat(
    'Duplikat wilayah-tahun',
    data_audit.duplicated(kunci, keep=False)
)

catat(
    'Tahun tidak valid',
    ~data_audit['tahun'].isin(tahun_target)
)

catat(
    'Kode wilayah kosong',
    data_audit['kode_kabupaten_kota'].isna()
)

# Cek seluruh kombinasi 514 wilayah x 3 tahun
wilayah = data_audit['kode_kabupaten_kota'].dropna().unique()

referensi = pd.MultiIndex.from_product(
    [wilayah, sorted(tahun_target)],
    names=kunci
)

tersedia = pd.MultiIndex.from_frame(
    data_audit[kunci].dropna()
)

kombinasi_hilang = referensi.difference(tersedia)

hasil_audit.append({
    'Pemeriksaan': 'Kombinasi wilayah-tahun lengkap',
    'Status': 'LULUS' if len(kombinasi_hilang) == 0 else 'GAGAL',
    'Jumlah Masalah': len(kombinasi_hilang),
    'Keterangan': 'Setiap wilayah memiliki 2023, 2024, 2025'
})


# ============================================================
# B. AUDIT MISSING VALUES
# ============================================================

print("B. AUDIT MISSING VALUES")

# Kolom YoY tahun 2023 memang belum memiliki pembanding 2022
kolom_yoy = [
    'perubahan_ipm_yoy',
    'perubahan_tpt_yoy',
    'perubahan_kemiskinan_yoy'
]

kolom_wajib = [
    c for c in data_audit.columns
    if c not in kolom_yoy
]

for col in kolom_wajib:
    catat(
        f'Missing value: {col}',
        data_audit[col].isna()
    )

# YoY wajib tersedia untuk 2024 dan 2025
for col in kolom_yoy:
    if col in data_audit.columns:
        catat(
            f'Missing YoY 2024-2025: {col}',
            data_audit['tahun'].isin([2024, 2025])
            & data_audit[col].isna()
        )


# ============================================================
# C. AUDIT TIPE DATA
# ============================================================

print("C. AUDIT TIPE DATA")

kolom_numerik = [
    'latitude',
    'longitude',
    'pertumbuhan_ekonomi',
    'pertumbuhan_produktivitas_tenaga_kerja',
    'persentase_pekerja_informal',
    'efisiensi_penggunaan_sumber_daya',
    'tingkat_pengangguran',
    'persentase_pemuda_neet',
    'persentase_pekerja_anak',
    'indeks_keselamatan_kerja',
    'kontribusi_lapangan_kerja_pariwisata',
    'tingkat_inklusi_keuangan',
    'indeks_pembangunan_manusia',
    'tingkat_kemiskinan',
    'tingkat_partisipasi_angkatan_kerja',
    'rata_rata_upah_bulanan',
    'jumlah_umkm',
    'persentase_akses_internet'
]

for col in kolom_numerik:
    nilai = pd.to_numeric(data_audit[col], errors='coerce')

    catat(
        f'Tipe numerik tidak valid: {col}',
        data_audit[col].notna() & nilai.isna()
    )


# ============================================================
# D. AUDIT RENTANG NILAI
# ============================================================

print("D. AUDIT RENTANG NILAI")

kolom_persentase = [
    'persentase_pekerja_informal',
    'tingkat_pengangguran',
    'persentase_pemuda_neet',
    'persentase_pekerja_anak',
    'tingkat_inklusi_keuangan',
    'tingkat_kemiskinan',
    'tingkat_partisipasi_angkatan_kerja',
    'persentase_akses_internet'
]

for col in kolom_persentase:
    nilai = pd.to_numeric(data_audit[col], errors='coerce')

    catat(
        f'Persentase di luar 0-100: {col}',
        nilai.notna() & ~nilai.between(0, 100)
    )

# IPM
nilai = pd.to_numeric(
    data_audit['indeks_pembangunan_manusia'],
    errors='coerce'
)

catat(
    'IPM di luar 0-100',
    nilai.notna() & ~nilai.between(0, 100)
)

# Nilai yang tidak boleh negatif
for col in ['rata_rata_upah_bulanan', 'jumlah_umkm']:
    nilai = pd.to_numeric(data_audit[col], errors='coerce')

    catat(
        f'Nilai negatif: {col}',
        nilai < 0
    )

# Jumlah UMKM harus bilangan bulat
umkm = pd.to_numeric(
    data_audit['jumlah_umkm'], errors='coerce'
)

catat(
    'Jumlah UMKM bukan bilangan bulat',
    umkm.notna() & ~np.isclose(umkm, np.round(umkm))
)


# ============================================================
# E. AUDIT KOORDINAT INDONESIA
# ============================================================

print("E. AUDIT KOORDINAT")

lat = pd.to_numeric(data_audit['latitude'], errors='coerce')
lon = pd.to_numeric(data_audit['longitude'], errors='coerce')

catat(
    'Latitude di luar Indonesia',
    lat.notna() & ~lat.between(-11, 6)
)

catat(
    'Longitude di luar Indonesia',
    lon.notna() & ~lon.between(95, 141.5)
)

# Koordinat satu wilayah seharusnya konsisten
for col in ['latitude', 'longitude']:
    jumlah_unik = data_audit.groupby(
        'kode_kabupaten_kota'
    )[col].transform('nunique')

    catat(
        f'Koordinat wilayah tidak konsisten: {col}',
        jumlah_unik > 1,
        tingkat='PERLU REVIEW'
    )


# ============================================================
# F. AUDIT KATEGORI
# ============================================================

print("F. AUDIT KATEGORI")

kategori_valid = {
    'jenis_kabupaten_kota': {
        'Kabupaten', 'Kota'
    },
    'kelompok_pulau': {
        'Sumatera',
        'Jawa',
        'Bali dan Nusa Tenggara',
        'Kalimantan',
        'Sulawesi',
        'Maluku',
        'Papua'
    },
    'kategori_urbanisasi': {
        'Perdesaan',
        'Peri-urban',
        'Perkotaan'
    }
}

for col, pilihan in kategori_valid.items():
    catat(
        f'Kategori tidak standar: {col}',
        data_audit[col].notna()
        & ~data_audit[col].isin(pilihan)
    )

# Penanda kategori tidak diketahui
kolom_kategori = [
    'sektor_ekonomi_dominan',
    'sektor_pekerjaan_dominan',
    'kategori_urbanisasi',
    'prioritas_kebijakan',
    'bantuan_perdagangan',
    'strategi_ketenagakerjaan_pemuda'
]

for col in kolom_kategori:
    nilai = (
        data_audit[col]
        .astype('string')
        .str.strip()
        .str.lower()
    )

    catat(
        f'Kategori placeholder: {col}',
        nilai.isin([
            'unknown', 'nan', 'none',
            'null', '?', '', 'tidak tersedia'
        ]),
        tingkat='PERLU REVIEW'
    )


# ============================================================
# G. AUDIT KONSISTENSI IDENTITAS WILAYAH
# ============================================================

print("G. AUDIT IDENTITAS WILAYAH")

kolom_identitas = [
    'kode_provinsi',
    'nama_provinsi',
    'nama_kabupaten_kota',
    'jenis_kabupaten_kota',
    'kelompok_pulau'
]

for col in kolom_identitas:
    jumlah_unik = data_audit.groupby(
        'kode_kabupaten_kota'
    )[col].transform('nunique')

    catat(
        f'Identitas wilayah berubah: {col}',
        jumlah_unik > 1
    )

# Satu kode provinsi tidak boleh memiliki nama berbeda
jumlah_nama_provinsi = data_audit.groupby(
    'kode_provinsi'
)['nama_provinsi'].transform('nunique')

catat(
    'Kode provinsi memiliki nama berbeda',
    jumlah_nama_provinsi > 1
)


# ============================================================
# H. AUDIT NILAI EKSTREM
# ============================================================

print("H. AUDIT OUTLIER")

# IQR hanya penanda statistik, bukan bukti data salah
kolom_outlier = [
    'pertumbuhan_ekonomi',
    'pertumbuhan_produktivitas_tenaga_kerja',
    'rata_rata_upah_bulanan',
    'jumlah_umkm',
    'tingkat_pengangguran',
    'tingkat_kemiskinan',
    'persentase_pemuda_neet'
]

for col in kolom_outlier:
    nilai = pd.to_numeric(data_audit[col], errors='coerce')

    q1 = nilai.quantile(0.25)
    q3 = nilai.quantile(0.75)
    iqr = q3 - q1

    if pd.notna(iqr) and iqr > 0:
        bawah = q1 - 1.5 * iqr
        atas = q3 + 1.5 * iqr

        catat(
            f'Outlier IQR: {col}',
            nilai.notna() & (
                (nilai < bawah) | (nilai > atas)
            ),
            tingkat='PERLU REVIEW',
            keterangan='Outlier statistik; bukan otomatis salah'
        )


# ============================================================
# I. AUDIT DATA REKONSTRUKSI
# ============================================================

print("I. AUDIT REKONSTRUKSI")

if 'status_rekonstruksi' in data_audit.columns:

    status_valid = {
        'Data tersedia',
        'Baris rekonstruksi'
    }

    catat(
        'Status rekonstruksi tidak dikenal',
        ~data_audit['status_rekonstruksi'].isin(
            status_valid
        )
    )

    jumlah_rekonstruksi = (
        data_audit['status_rekonstruksi']
        .eq('Baris rekonstruksi')
        .sum()
    )

    print(
        'Jumlah baris rekonstruksi:',
        jumlah_rekonstruksi
    )


# ============================================================
# J. AUDIT FEATURE ENGINEERING (JIKA SUDAH ADA)
# ============================================================

print("J. AUDIT FEATURE ENGINEERING")

# Rentang skor risiko 0-1
if 'skor_risiko_sdm' in data_audit.columns:
    skor = pd.to_numeric(
        data_audit['skor_risiko_sdm'],
        errors='coerce'
    )

    catat(
        'Skor risiko SDM di luar 0-1',
        skor.notna() & ~skor.between(0, 1)
    )

# Verifikasi YoY = nilai tahun sekarang - tahun sebelumnya
pasangan_yoy = {
    'perubahan_ipm_yoy': 'indeks_pembangunan_manusia',
    'perubahan_tpt_yoy': 'tingkat_pengangguran',
    'perubahan_kemiskinan_yoy': 'tingkat_kemiskinan'
}

data_urut = data_audit.sort_values(
    ['kode_kabupaten_kota', 'tahun']
)

for kolom_yoy, kolom_asal in pasangan_yoy.items():

    if kolom_yoy not in data_urut.columns:
        continue

    nilai_sebelumnya = data_urut.groupby(
        'kode_kabupaten_kota'
    )[kolom_asal].shift(1)

    tahun_sebelumnya = data_urut.groupby(
        'kode_kabupaten_kota'
    )['tahun'].shift(1)

    nilai_seharusnya = (
        data_urut[kolom_asal] - nilai_sebelumnya
    )

    nilai_aktual = pd.to_numeric(
        data_urut[kolom_yoy], errors='coerce'
    )

    tahun_berurutan = (
        data_urut['tahun'] - tahun_sebelumnya == 1
    )

    berbeda = (
        tahun_berurutan
        & nilai_aktual.notna()
        & ~np.isclose(
            nilai_aktual,
            nilai_seharusnya,
            rtol=1e-6,
            atol=1e-6
        )
    )

    catat(
        f'Perhitungan YoY salah: {kolom_yoy}',
        berbeda.reindex(data_audit.index, fill_value=False)
    )


# ============================================================
# K. RINGKASAN HASIL AUDIT
# ============================================================

laporan_audit_final = pd.DataFrame(hasil_audit)

if detail_masalah:
    detail_audit_final = pd.concat(
        detail_masalah,
        ignore_index=True
    )
else:
    detail_audit_final = pd.DataFrame(
        columns=[
            'kode_kabupaten_kota',
            'tahun',
            'Baris DataFrame',
            'Masalah',
            'Tingkat'
        ]
    )

jumlah_gagal = (
    laporan_audit_final['Status'] == 'GAGAL'
).sum()

jumlah_review = (
    laporan_audit_final['Status'] == 'PERLU REVIEW'
).sum()

jumlah_lulus = (
    laporan_audit_final['Status'] == 'LULUS'
).sum()

print("\n" + "=" * 55)
print("RINGKASAN AUDIT DATA CLEANING")
print("=" * 55)

print("Jumlah baris:", len(data_audit))
print("Jumlah kolom:", len(data_audit.columns))
print("Jumlah wilayah:", data_audit[
    'kode_kabupaten_kota'
].nunique())

print("\nPemeriksaan LULUS:", jumlah_lulus)
print("Pemeriksaan GAGAL:", jumlah_gagal)
print("Pemeriksaan PERLU REVIEW:", jumlah_review)

print("\n=== LAPORAN AUDIT ===")
display(laporan_audit_final)

print("\n=== DETAIL BARIS BERMASALAH ===")
display(detail_audit_final.head(100))

print("\n=== KOMBINASI WILAYAH-TAHUN HILANG ===")
display(kombinasi_hilang.to_frame(index=False))

if jumlah_gagal > 0:
    print("\nSTATUS AKHIR: GAGAL")
    print("Perbaiki masalah sebelum Feature Engineering.")
elif jumlah_review > 0:
    print("\nSTATUS AKHIR: LULUS BERSYARAT")
    print("Tidak ada kegagalan aturan dasar.")
    print("Periksa temuan PERLU REVIEW.")
else:
    print("\nSTATUS AKHIR: LULUS")
    print("Seluruh pemeriksaan yang diterapkan lulus.")
````

````python
# ============================================================
# FINALISASI STATUS AUDIT
# PERLU REVIEW -> LULUS (DITERIMA SEMENTARA)
# ============================================================

mask_review = (
    laporan_audit_final['Status'] == 'PERLU REVIEW'
)

# Tambahkan keterangan sebelum mengubah status
laporan_audit_final.loc[
    mask_review, 'Keterangan'
] = 'Diterima sementara; outlier dipertahankan dan belum diverifikasi'

# Ubah status laporan
laporan_audit_final.loc[
    mask_review, 'Status'
] = 'LULUS'

# Sinkronkan detail audit
if 'detail_audit_final' in globals():
    detail_audit_final.loc[
        detail_audit_final['Tingkat'] == 'PERLU REVIEW',
        'Tingkat'
    ] = 'LULUS'

# Hitung ulang ringkasan
print("=== RINGKASAN AUDIT FINAL ===")

print("Jumlah baris:", len(df_clean))
print("Jumlah wilayah:",
      df_clean['kode_kabupaten_kota'].nunique())

print("\nPemeriksaan LULUS:",
      (laporan_audit_final['Status'] == 'LULUS').sum())

print("Pemeriksaan GAGAL:",
      (laporan_audit_final['Status'] == 'GAGAL').sum())

print("Pemeriksaan PERLU REVIEW:",
      (laporan_audit_final['Status'] == 'PERLU REVIEW').sum())

display(laporan_audit_final)

if (laporan_audit_final['Status'] == 'GAGAL').sum() == 0:
    print("\nSTATUS AKHIR: LULUS DENGAN CATATAN")
    print("Outlier dipertahankan tanpa koreksi.")
else:
    print("\nSTATUS AKHIR: MASIH ADA PEMERIKSAAN GAGAL")
````
